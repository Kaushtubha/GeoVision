# GeoVision: Dataset Exploration & Integrity Analysis

This notebook provides exploratory analysis, class distribution checks, and image inspections for the 4 core datasets used in **GeoVision**:
1. **EuroSAT** (Scene Classification & Metric-Learning Retrieval)
2. **NWPU VHR-10** (Optical Remote Sensing Object Detection)
3. **LandCover.ai** (High-Resolution Aerial Land-Cover Segmentation)
4. **LEVIR-CD** (Bi-temporal Building Change Detection)

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from geovision.config import GeoVisionConfig
from geovision.constants import RAW_DATA_DIR
from geovision.data.downloaders import download_dataset, verify_dataset_integrity
from geovision.data.statistics import calculate_dataset_stats
```

## 1. Load Master Configuration

```python
cfg = GeoVisionConfig.load_default()
print(f"Experiment: {cfg.tracking.experiment_name}")
print(f"Seed: {cfg.seed}")
```

## 2. Generate and Verify Smoke Datasets

```python
for ds in ["eurosat", "nwpu_vhr10", "landcover_ai", "levir_cd"]:
    p = download_dataset(ds, target_dir=RAW_DATA_DIR, smoke=True)
    rep = verify_dataset_integrity(ds, p)
    print(f"{ds}: {rep}")
```

## 3. Dataset Statistical Summary

```python
stats_eurosat = calculate_dataset_stats(RAW_DATA_DIR / 'EuroSAT', 'classification')
print('EuroSAT Stats:', stats_eurosat)

stats_nwpu = calculate_dataset_stats(RAW_DATA_DIR / 'NWPU_VHR10', 'detection')
print('NWPU Stats:', stats_nwpu)

stats_lc = calculate_dataset_stats(RAW_DATA_DIR / 'LandCoverAI', 'segmentation')
print('LandCover.ai Stats:', stats_lc)

stats_levir = calculate_dataset_stats(RAW_DATA_DIR / 'LEVIR_CD', 'change')
print('LEVIR-CD Stats:', stats_levir)
```
